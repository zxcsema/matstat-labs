# Равенство средних двух независимых выборок

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
x = [33, 33, 26, 30, 28, 32, 32, 28, 32, 33, 27, 30, 27, 24, 26, 33, 38, 43, 33, 24, 31, 25,
     18, 39, 29, 23, 22, 19, 35, 36, 21, 29, 24, 32, 28, 28, 35, 26, 30, 33, 18, 34, 27, 33,
     31, 29, 28, 22]

y = [32, 33, 31, 25, 31, 21, 35, 31, 37, 37, 24, 30, 21, 26, 37, 37, 26, 37, 34, 31, 27, 32,
     33, 22, 39, 32, 28, 34, 21, 33, 35, 29, 34, 27, 28, 28, 33, 23, 22, 26, 32, 24, 32, 28,
     33, 32, 35, 27, 26, 31, 24, 26]

x = np.array(x)
y = np.array(y)
len(x), len(y)
```

```python
m = len(x)
n = len(y)

mx = x.mean()
my = y.mean()
d1 = x.var(ddof=1)
d2 = y.var(ddof=1)

print('средние', round(mx, 6), round(my, 6))
print('дисперсии', round(d1, 6), round(d2, 6))
```

```python
d = ((m - 1) * d1 + (n - 1) * d2) / (m + n - 2)
t = (mx - my) / np.sqrt(d * (1 / m + 1 / n))
tkr = stats.t.ppf(0.975, m + n - 2)

print('объединенная дисперсия', round(d, 6))
print('t набл', round(t, 6))
print('t крит', round(tkr, 6))
```

```python
stats.ttest_ind(x, y)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(9, 3))
ax[0].hist(x, bins=8, edgecolor='black')
ax[0].set_title('выборка X')
ax[1].hist(y, bins=8, edgecolor='black')
ax[1].set_title('выборка Y')
plt.show()
```

|t| = 0.726 меньше критического 1.984, в критическую область не попадаем, гипотезу о равенстве средних не отвергаем.
